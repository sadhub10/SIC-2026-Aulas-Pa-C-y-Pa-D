# **Tipos de Datos**

```python
# 1. Tipo de dato: Entero (int)
# Se usa para números enteros, sin decimales.
edad = 25
print("Variable 'edad' es un", type(edad)) # Muestra el tipo de dato: <class 'int'>

# 2. Tipo de dato: Flotante (float)
# Se usa para números con decimales.
precio_producto = 19.99
print("Variable 'precio_producto' es un", type(precio_producto)) # Muestra el tipo de dato: <class 'float'>

# 3. Tipo de dato: Cadena de texto (str)
# Se usa para texto. Siempre va entre comillas (dobles o simples).
nombre_completo = "Ana García"
print("Variable 'nombre_completo' es un", type(nombre_completo)) # Muestra el tipo de dato: <class 'str'>

# 4. Tipo de dato: Booleano (bool)
# Se usa para valores lógicos: Verdadero o Falso.
es_mayor_de_edad = True
print("Variable 'es_mayor_de_edad' es un", type(es_mayor_de_edad)) # Muestra el tipo de dato: <class 'bool'>

 #5. Ejemplo de cómo los tipos de datos afectan las operaciones
 #Intenta sumar un número (int) y una cadena de texto (str).
 #Esto causará un error porque los tipos son incompatibles.
 #print(edad + nombre_completo) # Si descomentas esta línea, el programa fallará.

# Sin embargo, puedes unir dos cadenas de texto.
bienvenida = "Hola " + nombre_completo
print(bienvenida)
```

# **Operadores Matemáticos**

```python
# Definimos dos variables numéricas para usarlas en los ejemplos
a = 20
b = 5

# 1. Suma (+)
# Suma los valores de 'a' y 'b'.
resultado_suma = a + b
print(f"La suma de {a} y {b} es: {resultado_suma}")

# 2. Resta (-)
# Resta el valor de 'b' de 'a'.
resultado_resta = a - b
print(f"La resta de {a} y {b} es: {resultado_resta}")

# 3. Multiplicación (*)
# Multiplica 'a' por 'b'.
resultado_multiplicacion = a * b
print(f"La multiplicación de {a} y {b} es: {resultado_multiplicacion}")

# 4. División (/)
# Divide 'a' entre 'b'. El resultado siempre es un número flotante (con decimales).
resultado_division = a / b
print(f"La división de {a} y {b} es: {resultado_division}")

# 5. División entera (//)
# Divide 'a' entre 'b' y redondea el resultado hacia abajo, eliminando los decimales.
resultado_division_entera = a // b
print(f"La división entera de {a} y {b} es: {resultado_division_entera}")

# 6. Módulo (%)
# Devuelve el residuo de la división de 'a' entre 'b'.
# 15 dividido por 4 es 3, con un residuo de 3.
resultado_modulo = a % b
print(f"El módulo de {a} y {b} es: {resultado_modulo}")

# 7. Potenciación (**)
# Eleva 'a' a la potencia de 'b'. Es decir, a elevado a b.
resultado_potencia = a ** b
print(f"La potencia de {a} elevado a {b} es: {resultado_potencia}")
```

# **Listas**

```python
# 1. Crear una lista de frutas
frutas = ["manzana", "banana", "cereza", "naranja"]

print("Mi lista de frutas es:", frutas)

# 2. Acceder a un elemento por su posición (índice)
# Las posiciones en Python empiezan en 0.
# El primer elemento es el 0, el segundo es el 1, y así sucesivamente.
primera_fruta = frutas[0]
print("La primera fruta de la lista es:", primera_fruta)

# 3. Modificar un elemento
# Cambiamos "banana" por "kiwi".
frutas[1] = "kiwi"
print("Lista después de modificar el segundo elemento:", frutas)

# 4. Agregar un nuevo elemento al final de la lista
frutas.append("uva")
print("Lista después de agregar 'uva':", frutas)

# 5. Eliminar un elemento
# Eliminamos "cereza" de la lista.
frutas.remove("cereza")
print("Lista después de eliminar 'cereza':", frutas)

# 6. Obtener la cantidad de elementos en la lista
cantidad_frutas = len(frutas)
print("Ahora hay", cantidad_frutas, "frutas en la lista.")
```

# **Condicionales**

```python
# 1. Definimos una variable para nuestro ejemplo
temperatura = 28

# 2. Usamos una sentencia 'if' para verificar una condición
# Si la temperatura es mayor a 25, imprimimos un mensaje.
if temperatura > 25:
  print("¡Hace calor!")

# 3. Usamos 'if', 'elif' y 'else' para múltiples condiciones
# El programa revisa las condiciones en orden.
# Solo se ejecuta el primer bloque que sea verdadero.
if temperatura < 10:
  print("Hace mucho frío.")
elif temperatura >= 10 and temperatura <= 20:
  print("El clima es templado.")
else:
  print("Hace calor, es el momento perfecto para la playa.")

# 4. Otro ejemplo con una condición diferente
tiene_dinero = False

if tiene_dinero:
  print("Puedes comprar un carro.")
else:
  print("No tienes suficiente dinero.")
```

# Ejemplo: Compra en el Mercado

```python
# 1. Variables iniciales
dinero_disponible = 50
costo_manzanas = 15
costo_pan = 10
costo_leche = 20
costo_total = costo_manzanas + costo_pan + costo_leche
# 2. Imprimir el estado actual
print(f"Dinero disponible: ${dinero_disponible}")
print(f"Costo de la compra: ${costo_total}")

# 3. Evaluar la condición
# La condición 'dinero_disponible >= costo_total' devuelve un valor booleano (True o False)
puedo_comprar = (dinero_disponible >= costo_total)

print(f"\n¿Tengo suficiente dinero para la compra? {puedo_comprar}")

# 4. Tomar una decisión basada en la condición booleana
if puedo_comprar:
  print("¡Genial! Tienes suficiente dinero para hacer la compra.")
  dinero_restante = dinero_disponible - costo_total
  print(f"Te sobran: ${dinero_restante}")
else:
  print("¡Uf! No tienes suficiente dinero.")
  dinero_faltante = costo_total - dinero_disponible
  print(f"Te faltan: ${dinero_faltante}")
```

# **Bucles**

```python
# 1. Definimos una lista de países
paises = ["Argentina", "Brasil", "Chile", "Colombia", "México"]

# 2. Usamos un bucle 'for' para recorrer la lista
# La variable 'pais' toma el valor de cada elemento de la lista en cada repetición.
for pais in paises:
  print(f"Me gustaría visitar {pais}.")

# 3. Otro ejemplo: bucle 'for' con la función range()
# La función range(5) genera números del 0 al 4.
print("\nContando de 0 a 4:")
for numero in range(5):
  print(numero)
```
