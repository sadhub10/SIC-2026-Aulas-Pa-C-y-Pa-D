# Categorias

## Sabemos que existen nulos pero se debe a que es un dataset maestro con la data cruda, tenemos que separarla para poder darle valor

```python
print(df_maestro.columns.tolist())
print(df_maestro.info())
```

```python
df_categoria = df_maestro[df_maestro["tipo"] == "categoria"].copy()
```

```python
df_categoria
```

```python
df_categoria = df_categoria.dropna(axis=1, how="all")
df_categoria = df_categoria.drop(columns=["tipo","producto"],)
```

```python
df_categoria
```

```python
df_categoria.to_csv("../../../datos/procesados/canasta_basica/categorias.csv",index=False)
```
