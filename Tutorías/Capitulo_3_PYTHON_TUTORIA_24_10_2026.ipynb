```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def __eq__(self, other):
        return self.nombre == other.nombre and self.precio == other.precio

# Crear dos productos con el mismo nombre y precio

p1 = Producto("Laptop", 500)
p2 = Producto("Laptop", 500)

print(p1 == p2)
print(p1 is p2)
```

### respuesta

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

# Con la implementación del método eq eq
class ProductoConEq:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def __eq__(self, otro):
        return self.nombre == otro.nombre and self.precio == otro.precio

# 1. Sin el método __eq__
p1 = Producto("Laptop", 500)
p2 = Producto("Laptop", 500)
print("Sin __eq__:", p1 == p2)

# 2. Con el método __eq__
p3 = ProductoConEq("Laptop", 500)
p4 = ProductoConEq("Laptop", 500)
print("Con __eq__:", p3 == p4)
```

## **Ejercicio 1: Analizador de Datos con Funciones ( returnMúltiples)**

* Enunciado: Escribe una función llamada analizar_inventario(precios)que reciba una lista de precios de productos.Debe calcular y devolver tres valores:el precio más caro,el precio más barato y el valor promedio del inventario.

image.png

### respuesta

```python
# Código de la solución
def analizar_inventario(precios):
    if not precios:
        return 0, 0, 0

    mas_caro = max(precios)
    mas_barato = min(precios)
    promedio = sum(precios) / len(precios)

    return mas_caro, mas_barato, promedio

# Datos de prueba
inventario_tienda = [120.5, 45.0, 300.0, 15.5, 89.9]
caro, barato, prom = analizar_inventario(inventario_tienda)

# Resultados esperados
print(f"Precio máximo: {caro}")
print(f"Precio mínimo: {barato}")
print(f"Precio promedio: {prom:.2f}")
```

## **Ejercicio 2: Generador de Tasas (Aplicación de Cierres)**

* Enunciado: Diseña una función de llamada de fábrica crear_conversor(tasa_cambio)que restablecer como cierre.Debe retornar una función interna que convierta una cantidad de dinero local en otra moneda multiplicándola por la tasa establecida.

### respuesta

```python
# Código de la solución
def crear_conversor(tasa_cambio):
    def convertir(monto):
        return monto * tasa_cambio
    return convertir

# Datos de prueba (ej: Tasa de cambio de moneda A a moneda B = 36.5)
convertir_a_dolares = crear_conversor(36.5)

# Resultados esperados
print(convertir_a_dolares(10))   # 10 unidades
print(convertir_a_dolares(100))  # 100 unidades
```

## Ejercicio 3: Control de Empleados (POO y Métodos Dunder)

* Enunciado: Crea una clase Empleadoque recibe nombrey sueldo.

Implementa __str__para que al imprimir el objeto devuelva: . Empleado: [Nombre] - Sueldo: $[Sueldo]

Implementa __add__para permitir sumar los sueldos de dos empleados utilizando directamente el operador +.

### respuesta

```python
# Código de la solución
class Empleado:
    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def __str__(self):
        return f"Empleado: {self.nombre} - Sueldo: ${self.sueldo}"

    def __add__(self, otro):
        # Suma los sueldos y retorna un total numérico
        return self.sueldo + otro.sueldo

# Datos de prueba
emp1 = Empleado("Ana", 1200)
emp2 = Empleado("Carlos", 1500)

# Probando __str__
print(emp1)
print(emp2)

# Probando __add__
sueldo_total = emp1 + emp2
print(f"Suma total de sueldos: ${sueldo_total}")
```
